ora qui mi creo il dataset =>  Il dataset dovrà essere suddiviso in tre sottoinsiemi: un training set con almeno
1000 paragrafi per ciascun autore, e un validation set e un test set composti da
almeno 100 paragrafi per autore ciascuno. È fondamentale che i libri utilizzati
per la fase di addestramento non siano presenti nei set di validazione e test

```python
import json
import os
import random
from collections import defaultdict
```

```python
dizionario = json.load(open("/content/drive/MyDrive/LC2 progetto/paragrafi_annidati.json"))
```

```python
# ora vado a togliere e salvarmi in nuovi dizionari i testi che voglio tenermi da parte per il test e il
# validation, gli altri invece li uso per crearmi il training

test_dickens = {
    "Dickens David": dizionario["dickens"].pop("Dickens David")
}

test_austen = {
    "Austen Pride": dizionario["austen"].pop("Austen Pride")
}

test_woolf = {
    "Woolf Mrs": dizionario["woolf"].pop("Woolf Mrs"),
    "Woolf Common": dizionario["woolf"].pop("Woolf Common"),
    "Woolf Monday": dizionario["woolf"].pop("Woolf Monday")
}

# dizionario
```

```python
# ora devo creare un file per ogni paragrafo di ogni autore
# in totale 3000 file quindi
# faccio questa operazione sul dizionario a cui ho appena tolto i libri per il test


for autore, libri in dizionario.items():
  count = 1
  # e poi essendo che sono annidati
  for titolo, paragrafi in libri.items():
    for paragrafo in paragrafi:
      #quindi per ogni paragrafo:
      if count <= 1000:
        nome_file = f"{autore}_{count}_training.txt"
        file = open(f"/content/drive/MyDrive/LC2 progetto/dataset/training/{nome_file}", "w")
        file.write(paragrafo)
        file.close()
      count += 1
```
