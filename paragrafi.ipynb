```python
import os
import json
import nltk
from collections import defaultdict

nltk.download('punkt')
nltk.download('punkt_tab', quiet=True, force=True) # Download the punkt_tab resource
```

 I paragrafi validi per l’analisi dovranno contenere un numero di
token compreso tra 50 e 100, dove i token possono essere approssimati
dividendo il testo sugli spazi, senza ricorrere a una tokenizzazione linguistica
avanzata


```python
#prima di tutti mi prendo i file che devo andare a vedere che sono quelli cehe ho normalizzato prima
percorso_testi = "/content/drive/MyDrive/LC2 progetto/testi"
lista_file = []
for cartella, sottocartelle, files in os.walk(percorso_testi):
    for file in files:
        if file.endswith("_normalizzato.txt"):
            file_path = os.path.join(cartella, file)
            lista_file.append(file_path)

print(lista_file)
```

```python
#ora per ogni file vado a prendere i paragrafi e se sono tra i 50 e i 100 token li tengo

# definisco il dizionario
dizionario = defaultdict(lambda: defaultdict(list))  # autore → libro → [paragrafi]

# per ogni file nella lista dei normalizzati di prima vadi a leggere e dividere in paragrafi
for file_path in lista_file:
    with open(file_path, "r", encoding="utf-8") as f:
        contenuto = f.read()
        paragrafi = contenuto.split("\n\n")

        for paragrafo in paragrafi:
            paragrafo = paragrafo.strip() # pulisco spazi a inizio o fine
            if not paragrafo: # ignoro i èaragrafi vuoti
                continue

            tokens = nltk.word_tokenize(paragrafo)
            num_token = len(tokens) # conto quanti token ci sono nel paragrafo

            if 50 <= num_token <= 100:
                nome_file = os.path.basename(file_path)
                # Ricava autore e libro
                autore = nome_file.split(" ")[0].lower()
                nome_libro = nome_file.replace("_normalizzato.txt", "")

                # Aggiunge il paragrafo al dizionario
                dizionario[autore][nome_libro].append(paragrafo)

# Salva dizionario
output_dict_path = "/content/drive/MyDrive/LC2 progetto/paragrafi_annidati.json"
with open(output_dict_path, "w", encoding="utf-8") as out_file:
    json.dump(dizionario, out_file, ensure_ascii=False, indent=2)

# print(f"Salvati {len(paragrafi_ok)} paragrafi in:\n{output_path}")
print(f"Salvato dizionario annidato in:\n{output_dict_path}")
```
