```python
import re
import os
```

```python
percorso_testi = "/content/drive/MyDrive/LC2 progetto/testi"
lista_file = []
for cartella, sottocartelle, files in os.walk(percorso_testi):
    for file in files:
        if file.endswith(".txt"):
            file_path = os.path.join(cartella, file)
            lista_file.append(file_path)

print(lista_file)
```

Si raccomanda particolare attenzione nella creazione dei paragrafi: la maggior parte dei
file .txt dei libri disponibili sul Progetto Gutenberg presenta interruzioni di riga anche all'interno
di frasi, il che può compromettere l'analisi testuale. Per evitare tali problemi, è necessario
unificare ciascun paragrafo sostituendo tutti i ritorni a capo interni con uno spazio.


```python
def funzione_normalizzazione(file_originale, file_normalizzato):

    # regex per andare a sostituire tutti gli \n che sono da soli (quindi non hanno un \n prima o dopo)
    pattern = "(?<!\n)\n(?!\n)"
    repl = " "

    f = open(file_originale)
    stringa = f.read()

    stringa_mod = re.sub(pattern, repl, stringa)

    # dove ho 3 \n consecutivi metto due \n
    pattern2 = "\n{3,}"
    repl2 = "\n\n"

    stringa_mod = re.sub(pattern2, repl2, stringa_mod)

    # Scrivo la stringa modificata su un nuovo file
    f_norm = open(file_normalizzato, "w")
    f_norm.write(stringa_mod)

    f_norm.close()
    f.close()
```

```python
# normalizzo tutti i file

for file in lista_file:
    file_originale = file
    file_normalizzato = file.split('.')[0]+"_normalizzato.txt"

    funzione_normalizzazione(file_originale, file_normalizzato)
```
